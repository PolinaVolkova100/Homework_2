```python
## 7. Написать код для создания срезов массива 3 на 4
c = np.random.randint(10, size=(3, 4))
print(c)
## - первые две строки и три столбца
print(c[:2, :3])
## - первые три строки и второй столбец
print(c[::, 1])
## - все строки и столбцы в обратном порядке
print(c[::-1, ::-1])
## - второй столбец
print(c[::, 1])
## - третья строка
print(c[2, ::])
```

```python
## 9. Продемонстрируйте использование newaxis для получения вектора-столбца и вектора-строки
a = np.arange(1, 10)
a1 = a[:, np.newaxis]
a2 = a[np.newaxis, :]

print(a)
print(a1)
print(a2)
```

```python
## 10. Разберитесь, как работает метод dstack
# Используется для стека массивов по глубине (вдоль третьей оси)
x = np.array([1, 2, 3])
y = np.array([4, 5, 6])
r1 = np.vstack([x, y])
r2 = np.dstack([r1, r1])
print(r2)
```

```python
## 11. Разберитесь, как работают методы split, vsplit, hsplit, dsplit
# numpy.split() позволяет разделить массив на несколько подмассивов вдоль заданной оси
# Синтаксис: numpy.split(array, indices_or_sections, axis=0)
a = np.arange(10)
print(np.split(a, 5)) # деление на 5 подмассивов по строкам
print(np.split(a, 2)) # деление на 2 подмассива по строкам
print(np.split(a, [3, 5])) # деление на подмассивы с указанием массива, разбивающего исходный массив

b = np.random.randint(10, size=(6, 4))
print(b)
# vsplit разбивает массив на несколько подмассивов по вертикали (строчно)
print(np.vsplit(b, 3))
print(np.vsplit(b, 2))

# hsplit разделяет массив на несколько подмассивов горизонтально (по столбцам)
c = np.random.randint(10, size=(4, 6))
print(np.hsplit(c, 3))
print(np.hsplit(c, 2))

# dsplit разделяет массив на несколько подмассивов вдоль третьей оси (глубины)
d = np.random.randint(10, size=(4, 5, 6))
print(np.dsplit(d, 3))
print(np.dsplit(d, 2))
```

```python
## 12. Привести пример использования всех универсальных функций, которые я привел

a = np.array([[1, 2, 3], [2, 3, 4], [3, 4, 5]])
b = np.array([[0, 0, 3], [7, 5, 4], [7, 8, 9]])
# вычитание массивов
print(a - b)

# -
print(-a) 
print(-b)

# деление
print(b / a)
print(b / 2)

# целочисленное деление
print(b // a)
print(b // 2)

# Возведение в степень
print(a ** 3)
print(b ** a)

# остаток от джеления
print(b % 3)
print(b % a)
```
