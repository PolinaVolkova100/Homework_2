```python
import pandas as pd
import numpy as np
```

```python
# # 1. Разобраться как использовать мультииндексные ключи в данном примере
index = [
    ('city_1', 2010),
    ('city_1', 2020),
    ('city_2', 2010),
    ('city_2', 2020),
    ('city_3', 2010),
    ('city_3', 2020),
]
index = pd.MultiIndex.from_tuples(index)

population = [
    101,
    201,
    102,
    202,
    103,
    203,
]
pop = pd.Series(population, index = index)
pop_df = pd.DataFrame(
    {
        'total': pop,
        'something': [
            10,
            11,
            12,
            13,
            14,
            15,
        ]
    }
)
print(pop_df)
```

```python
pop_df_1 = pop_df.loc['city_1', 'something']
print(pop_df_1)
```

```python
pop_df_2 = pop_df.loc[['city_1', 'city_3'], ['total', 'something']]
print(pop_df_2)
```

```python
pop_df_3 = pop_df.loc[['city_1', 'city_3'], 'something']
print(pop_df_3)
```

```python
index = pd.MultiIndex.from_product(
    [
        ['city_1', 'city_2'],
        [2010, 2020]
    ],
    names=['city', 'year']
)

columns = pd.MultiIndex.from_product(
    [
        ['person_1', 'person_2', 'person_3'],
        ['job_1', 'job_2']
    ],
    names=['worker', 'job']
)

rng = np.random.default_rng(1)
data = rng.random((4, 6))
data_df = pd.DataFrame(data, index=index, columns=columns)
print(data_df)
```

```python
# 2. Из получившихся данных выбрать данные по 
# - 2020 году (для всех столбцов)
print(data_df.loc[pd.IndexSlice[:, 2020],:])
```

```python
# - job_1 (для всех строк)
print(data_df.loc[:, pd.IndexSlice[:, 'job_1']])
```

```python
# - для city_1 и job_2 
print(data_df.loc[pd.IndexSlice['city_1', :], pd.IndexSlice[:, 'job_2']])
```

```python
# - все данные по первому городу и первым двум person-ам (с использование срезов)
print(data_df.loc[pd.IndexSlice['city_1', :], pd.IndexSlice[('person_1', 'person_2'), :]])
```

```python
# Приведите пример (самостоятельно) с использованием pd.IndexSlice
print(data_df.loc[pd.IndexSlice['city_2', 2010], pd.IndexSlice[('person_1', 'person_3'), 'job_1']])
```

```python
#4. Привести пример использования inner и outer джойнов для Series (данные примера скорее всего нужно изменить)
ser1 = pd.Series(['a', 'b', 'c'], index=[1,2,3])
ser2 = pd.Series(['b', 'c', 'f'], index=[2,3,4])

print (pd.concat([ser1, ser2], join='outer', axis=1))
print()
print (pd.concat([ser1, ser2], join='inner', axis=1))
```
