```python
import numpy as np 
import matplotlib as mpl 
import matplotlib.pyplot as plt
import numpy.ma as ma 
```

```python
x = np.linspace(0, 21, 5)
y1 = np.array([0, 7, 3, 4, 11])
y2 =np.array([4, 3, 0, 8, 12])

plt.figure()
plt.plot(x, y1, color='red', marker='o', label='line 1')
plt.plot(x, y2, color='green', marker='o', linestyle='dashed', label='line 2')
plt.legend()
plt.show()
```

```python
x = np.arange(1, 5.5, 0.5)
y1 = np.array([1, 4, 7, 6.5, 6, 4.5, 3, 3.5, 4])
y2 = np.array([9, 6.5, 4, 3, 2, 3, 4, 6.5, 9])
y3 = np.array([-7, -5.5, -4, -1, 2, -1, -4, -5.5, -7])

plt.figure()
plt.subplot(2, 2, (1, 2))
plt.plot(x, y1)

plt.subplot(2, 2, 3)
plt.plot(x, y2)

plt.subplot(2, 2, 4)
plt.plot(x, y3)

plt.show()
```

```python
fig, ax = plt.subplots()
x = np.linspace(-5, 5, 15) 
y = x ** 2 
ax.plot(x, y) 
ax.annotate('min', xy = (0, 0), xytext=(0, 9), arrowprops=dict(facecolor='green'))

plt.show()
```

```python
x = np.random.randint(0, 8, 100)
y = np.random.randint(0, 8, 100)

plt.hist2d(x, y, bins=7)

plt.colorbar()

plt.show()
```

```python
x = np.linspace(0, 5, 100) 
y = np.cos(3 * x)
plt.plot(x, y, color='red')
plt.fill_between(x, y, color='royalblue')

plt.show()
```

```python
x = np.linspace(0, 5, 500) 
y = np.cos(3 * x)
y = ma.masked_where(y < -0.5, y) 
plt.plot(x, y)

plt.show()
```

```python
plt.subplots(figsize=(18, 6))

x = np.arange(0, 7)

plt.subplot(1, 3, 1)
plt.step(x, x, '-go')
plt.grid()

plt.subplot(1, 3, 2)
plt.step(x, x, '-go', where='post')
plt.grid()

plt.subplot(1, 3, 3)
plt.step(x, x, '-go', where='mid')
plt.grid()

plt.show()
```

```python
x = np.linspace(0, 10, 100) 

y1 = -0.16 * x ** 2 + 1.6 * x
y2 = -0.56 * x ** 2 + 5.6 * x
y3 = -0.53 * x ** 2 + 7.42 * x

plt.fill_between(x, y1, color='mediumslateblue', label='y1')
plt.fill_between(x, y2, y1, color='darkorange', label='y2')
plt.fill_between(x, y3, y2, color='forestgreen', label='y3')
plt.legend()

plt.show()
```

```python
cars = ['Ford', 'Toyota', 'BMV', 'AUDI', 'Jaguar']

data = [13, 12, 34, 16, 25]
explode = [0, 0, 0.1, 0, 0]

fig = plt.figure(figsize=(10, 7))
plt.pie(data, labels=cars, explode = explode)

plt.show()
```

```python
cars = ['Ford', 'Toyota', 'BMV', 'AUDI', 'Jaguar']
data = [13, 12, 34, 16, 25]

fig, ax = plt.subplots()
ax.pie(data, labels=cars, wedgeprops=dict(width=0.5))

plt.show()
```
