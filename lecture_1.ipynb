```python
import numpy as np
import sys
import array
```

```python
# Типы данных в Python
# Динамическая типизация - тип данных переменной определяется в ходе выполнения кода

x = 1
print(type(x))  # <class 'int'>
print(sys.getsizeof(x)) # 28 (байт)

x = 'hello'
print(type(x))  # <class 'str'>

x = True
print(type(x)) # <class 'bool'>

l1 = list([])
print(sys.getsizeof(l1)) # 56

l2 = list([1, 2, 3]) # В листе можно хранить элементы разных типов
print(sys.getsizeof(l2))  # 88
```

```python
a1 = array.array('i', [1, 2, 3]) # масссив - можно хранить элементы только одного типа
# 'i' - код типа int
print(sys.getsizeof(a1))
print(type(a1))  
```

```python
a = np.array([1, 2, 3, 4, 5])
print(type(a), a) # <class 'numpy.ndarray'>
```

```python
# "повышающее" приведение типов
a = np.array([1.23, 2, 3, 4, 5])
print(type(a), a)
```

```python
# параметр dtype для указания типа
a = np.array([1.23, 2, 3, 4, 5], dtype=int)
print(type(a), a)
```

```python
# создание многомерного массива
a = np.array([range(i, i + 3) for i in [2, 4, 6]])
print(a, type(a))
```

```python
# массив из нулей
a = np.zeros(10, dtype=int)
print(a, type(a))
```

```python
# многомерный массив из единиц
a = np.ones((3, 5), dtype=float)
print(a, type(a))
```

```python
# массив с определенными значениями
print(np.full((4,5), 3.1415))
```

```python
print(np.arange(0, 20, 2))
```

```python
# получение единичной матрицы определенного размера
print(np.eye(4))
```

```python
x1 = np.random.randint(10, size=3)
x2 = np.random.randint(10, size=(3, 2))
x3 = np.random.randint(10, size=(3, 2, 1))

# print(x1)
# print(x2)
# print(x3)

# Число размерностей массива, размер каждой размерности и общий размер массива
print(x1.ndim, x1.shape, x1.size)
print(x2.ndim, x2.shape, x2.size)
print(x3.ndim, x3.shape, x3.size)
```

```python
# Индекс (с 0)
a = np.array([1, 2, 3, 4, 5])
print(a[0])
print(a[-2])
a[1] = 20
print(a)
```

```python
# доступ по индексам в многомерных массивах
a = np.array([[1, 2], [3, 4]])
print(a)
print(a[0, 0])
print(a[-1, -1])
a[1, 0] = 100
print(a)
```

```python
a = np.array([1, 2, 3, 4])
b = np.array([1.0, 2, 3, 4])

print(a)
print(b)

a[0] = 10
print(a)

a[0] = 10.123
print(a)
```

```python
## Срез - подмассив [start:finish:step], по умолчанию - [0:shape:1]
a = np.array([1, 2, 3, 4, 5, 6])
print(a[:3])
print(a[3:])
print(a[1:5])
print(a[1:-1])
print(a[1::2])
print(a[::1])
print(a[::-1])
```

```python
# срезы - не копии, а ссылки на части массива
b = a[:3]
print(b)

b[0] = 100
print(a)
```

```python
a = np.arange(1, 13)
print(a)
# меняем размерность массива
print(a.reshape(2, 6))
print(a.reshape(3, 4))
```

```python
x = np.array([1, 2, 3])
y = np.array([4, 5])
z = np.array([6])

# объединение массивов
print(np.concatenate([x, y, z]))

x = np.array([1, 2, 3])
y = np.array([4, 5, 6])

# вертикальное склеивание
r1 = np.vstack([x, y])
print(r1)

# горизонтальное склеивание
print(np.hstack([r1, r1]))
```
