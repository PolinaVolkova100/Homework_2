```python
import numpy as np
import pandas as pd
```

```python
# Pandas - расширение NumPy (структурированные массивы). Строки и столбцы индексируются метками, а не только числовыми значениями
```

```python
# Основные структуры данных в Pandas
# Series, DataFrame, Index
```

```python
## Series
data = pd.Series([0.25, 0.5, 0.75, 1.0])
print(data)
print(type(data))

print(data.values)
print(type(data.values))

print(data.index)
print(type(data.index))
```

```python
print(data[0])
print(data[1:3])
```

```python
data = pd.Series([0.25, 0.5, 0.75, 1.0], index=['a', 'b', 'c', 'd'])
print(data)
print(data['a'])
print(data['b':'d'])
```

```python
print(type(data.index))
```

```python
data = pd.Series([0.25, 0.5, 0.75, 1.0], index=[1, 10, 7, 'd']) # тип индексов может быть разным
print(data)
print(data[1])
print(data[10:'d'])
```

```python
population_dict = {
    'city_1': 1001,
    'city_2': 1002,
    'city_3': 1003,
    'city_4': 1004,
    'city_5': 1005,
}
population = pd.Series(population_dict) #можно собирать Series на основе словаря
print(population)
print(population['city_4'])
print(population['city_4':'city_5'])
```

```python
# Для создания Series можно использовать
#1. Списки Python или массивы NumPy
#2. Скалярные значения
#3. Словари
```

```python
## DataFrame - двумерный массив с явно определенными индексами. Последовательность "согласованных" объектов Series
population_dict = {
    'city_1': 1001,
    'city_2': 1002,
    'city_3': 1003,
    'city_4': 1004,
    'city_5': 1005,
}

area_dict = {
    'city_1': 9991,
    'city_2': 9992,
    'city_3': 9993,
    'city_4': 9994,
    'city_5': 9995,
}

population = pd.Series(population_dict)
area = pd.Series(area_dict)
print(population)
print(area)
```

```python
states = pd.DataFrame({
    'population': population,
    'area': area,
})
print(states)
```

```python
print(states.values)
print(states.index)
print(states.columns)

print(type(states.values))
print(type(states.index))
print(type(states.columns))
```

```python
print(states['area'])
```

```python
# DataFrame. Способы создания
#1. Через объекты Series
#2. Списки словарей
#3. Cловари объектов Series
#4. Двумерный массив NumPy
#5. Структурированный массив NumPy
```

```python
# Index - способ организации ссылки на данные объектов Series и DataFrame. Index - неизменяем, упорядочен, является
# мультимножеством (могут быть повторяющиеся значения)
ind = pd.Index([2, 3, 5, 7, 11])
print(ind[1])
print(ind[::2])
#ind[1] = 5 - возникнет ошибка, т.к. объект неизменяем
```

```python
#Index - следует соглашениям объекта set
indA = pd.Index([1, 2, 3, 4, 5])
indB = pd.Index([2, 3, 4, 5, 6])
print(indA.intersection(indB))
```

```python
# Выборка данных из Series
## как словарь
data = pd.Series([0.25, 0.5, 0.75, 1.0], index=['a', 'b', 'c', 'd'])
print('a' in data)
print('z' in data)
print(data.keys())
print(list(data.items()))
data['a'] = 100
data['z'] = 1000
print(data)
```

```python
## как одномерный массив
data = pd.Series([0.25, 0.5, 0.75, 1.0], index=['a', 'b', 'c', 'd'])
print(data['a':'c'])
print(data[0:2])
print(data[(data > 0.5) & (data < 1)])
print(data[['a', 'd']])
```

```python
# атрибуты-индексаторы
data = pd.Series([0.25, 0.5, 0.75, 1.0], index=[1, 3, 10, 15])
print(data[1])
print(data.loc[1])
print(data.iloc[1])
```

```python
# выборка данных из DataFrame
## как словарь
pop = pd.Series(
    {
        'city_1': 1001,
        'city_2': 1002,
        'city_3': 1003,
        'city_4': 1004,
        'city_5': 1005,
    })

area = pd.Series(
    {   'city_1': 9991,
        'city_2': 9992,
        'city_3': 9993,
        'city_4': 9994,
        'city_5': 9995,
    })

data = pd.DataFrame({'area1': area, 'pop1': pop, 'pop': pop})
print(data)
print(data['area1'])
print(data.area1)
```

```python
print(data.pop1 is data['pop1'])
print(data.pop is data['pop'])
```

```python
data['new'] = data['area1']
data['new1'] = data['area1'] / data['pop1']
print(data)
```

```python
## как двумерный NumPy-массив
data = pd.DataFrame({'area1': area, 'pop1': pop})
print(data)
print(data.values)
print(data.T)
print(data['area1'])
print(data.values[0:3])
```

```python
# атрибуты-индексаторы
data = pd.DataFrame({'area1': area, 'pop1': pop, 'pop': pop})

print(data)
print(data.iloc[:3, 1:2])
print(data.loc[:'city_4', 'pop1':'pop'])
print(data.loc[data['pop'] > 1002, ['area1', 'pop']])
```

```python
data.iloc[0, 2] = 999999
print(data)
```

```python
print(df - df.iloc[0])
```

```python
print(df.iloc[0, ::2])
print(df - df.iloc[0, ::2])
```

```python
# NaN - not a number
# NA - значения: NaN, null, -99999
# Pandas. Два способа хранения отсутствующих значений
# индикаторы NaN, None
# null

# None - объект (накладные расходы). Не работает с sum, min

val1 = np.array([1, 2, 3])
print(val1.sum())

# val1 = np.array([1, Nope, 2, 3])
# print(val1.sum()) - вызовет ошидку

val1 = np.array([1, np.nan, 2, 3])
print(val1)
print(val1.sum())
print(np.nansum(val1))
```

```python
x = pd.Series(range(10), dtype=int)
print(x)
```

```python
x[0] = None
x[1] = np.nan
print(x)
```

```python
x1 = pd.Series(['a', 'b', 'c'])
print(x1)
x[0] = None
x[1] = np.nan
print(x1)
```

```python
x2 = pd.Series([1, 2, 3, np.nan, None, pd.NA])
print(x2)
```

```python
x3 = pd.Series([1, 2, 3, np.nan, None, pd.NA], dtype='Int32')
print(x3)
```

```python
print(x3.isnull())
print(x3[x3.isnull()])
print(x3[x3.notnull()])
```

```python
print(x3.dropna())
```

```python
df = pd.DataFrame([[1, 2, 3, np.nan, None, pd.NA], [1, 2, 3, 4, 5, 6], [1, np.nan, 3, 4, np.nan, 6]])
print(df)
print()
print(df.dropna())
print()
print(df.dropna(axis=0))
print()
print(df.dropna(axis=1))
```

```python
# how
# - all - все значения NA
print(df.dropna(axis=1, how='all'))
# - any - хотя бы одно значение
print(df.dropna(axis=1, how='any'))
# - thresh = x, остается, если присутствует МИНИМУМ х НЕПУСТЫХ значений
print(df.dropna(axis=1, thresh=2))
```
