```python
import pandas as pd
import numpy as np
```

```python
# Если размерность данных > 2, то используют иерархическую индексацию (мультииндекс). В один индекс включается несколько уровней.
index = [
    ('city_1', 2010),
    ('city_1', 2020),
    ('city_2', 2010),
    ('city_2', 2020),
    ('city_3', 2010),
    ('city_3', 2020),
]

population = [
    101,
    201,
    102,
    202,
    103,
    203,
]

pop = pd.Series(population, index=index)

print(pop)
```

```python
print(pop[[i for i in pop.index if i[1] == 2020]])
```

```python
# MultiIndex

index = pd.MultiIndex.from_tuples(index)

pop = pop.reindex(index)

print(pop)
```

```python
print(pop[:, 2020])
```

```python
pop_df = pop.unstack() # преобразование мультииндексных Series в DataFrame
print(pop_df)
```

```python
print(pop_df.stack()) # преобразование мультииндексных DataFrame в Series
```

```python
### Как можно создавать мультииндексы
### - список массивов, задающих значение индекса на каждом уровне
i1 = pd.MultiIndex.from_arrays(
    [
        ['a', 'a', 'b', 'b'],
        [1, 2, 1, 2]
    ]
)
print(i1)
```

```python
# - список кортежей, задающих значение индекса в каждой точке
i2 = pd.MultiIndex.from_tuples(
    [
        ('a', 1),
        ('a', 2),
        ('b', 1),
        ('b', 2),
    ]
)

print(i2)
```

```python
# - декартово произведение обычных индексов
i3 = pd.MultiIndex.from_product(
    [
        ['a', 'b'],
        [1, 2]
    ]
)
print(i3)
```

```python
# - описание внутреннего представления: levels, codes
i4 = pd.MultiIndex(
    levels = [
        ['a', 'b', 'c'],
        [1, 2]
    ],
    codes = [
        [0, 0, 1, 1, 2, 2], # a a b b c c
        [0, 1, 0, 1, 0, 1], # 1 2 1 2 1 2
    ]
)
print(i4)
```

```python
# Перегруппировка мультииндексов
rng = np.random.default_rng(1)

index = pd.MultiIndex.from_product(
    [
        ['a', 'b', 'c'],
        [1, 2]
    ]
)

data = pd.Series(rng.random(6), index=index)
data.index.names = ['char', 'int']

print(data)
print(data['a':'b'])
```

```python
data = data.sort_index()
print(data)
print(data['a':'b'])
```

```python
index = [
    ('city_1', 2010, 1),
    ('city_1', 2010, 2),
    
    ('city_1', 2020, 1),
    ('city_1', 2020, 2),
    
    ('city_2', 2010, 1),
    ('city_2', 2010, 2),
    
    ('city_2', 2020, 1),
    ('city_2', 2020, 2),
    
    ('city_3', 2010, 1),
    ('city_3', 2010, 2),
    
    ('city_3', 2020, 1),
    ('city_3', 2020, 2), 
]

population = [
    101,
    1010,
    201,
    2010,
    102,
    1020,
    202,
    2020,
    103,
    1030,
    203,
    2030, 
]

pop = pd.Series(population, index=index)
print(pop)
```

```python
# NumPy Конкатенация
x = [1, 2, 3]
y = [4, 5, 6]
z = [7, 8, 9]

print(np.concatenate([x, y, z]))
```

```python
x = [[1, 2, 3]]
y = [[4, 5, 6]]
z = [[7, 8, 9]]
print(np.concatenate([x, y, z]))
```

```python
print(np.concatenate([x, y, z], axis=1))
```

```python
# Pandas - concat
ser1 = pd.Series(['a', 'b', 'c'], index=[1, 2, 3])
ser2 = pd.Series(['d', 'e', 'f'], index=[4, 5, 6])
print(pd.concat([ser1, ser2]))
```

```python
ser1 = pd.Series(['a', 'b', 'c'], index=[1, 2, 3])
ser2 = pd.Series(['b', 'c', 'f'], index=[4, 5, 6])
print(pd.concat([ser1, ser2], verify_integrity=False))
```

```python
print(pd.concat([ser1, ser2], ignore_index=True))
```

```python
print(pd.concat([ser1, ser2], keys=['x', 'y']))
```

```python
print(pd.concat([ser1, ser2], join='outer'))
print(pd.concat([ser1, ser2], join='inner'))
```
