```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# I
np.random.seed(1)
x = 10 * np.random.rand(50)
y = 2 * x - 1 + np.random.randn(50)

plt.scatter(x, y)
plt.show()
```

```python
# II
# 1. Выбираем класс модели
from sklearn.linear_model import LinearRegression

#2. Устанавливаем гиперпараметры модели
model = LinearRegression(fit_intercept=True)
```

```python
#3. Создаем матрицу признаков и целевой массив
print(x.shape)
print(y.shape)

X = x[:, np.newaxis]
```

```python
#4. Обучение модели fit()
model.fit(X, y)
# коэффициент а
print(model.coef_[0])
# коэффициент b
print(model.intercept_)
```

```python
x_ = np.linspace(0, 10, 30)
y_ = model.coef_[0] * x_ + model.intercept_

plt.scatter(x, y)
plt.plot(x_, y_)
plt.show()
```

```python
#5. Применять модель к новым данным
xfit = np.linspace(0, 10, 5)
yfit = model.predict(xfit[:, np.newaxis])

plt.scatter(x, y)
plt.plot(x_, y_)
plt.scatter(xfit, yfit)
plt.show()
```
